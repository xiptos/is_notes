# The Gymnasium loop

[Gymnasium](https://gymnasium.farama.org/) provides environments for reinforcement learning. The interaction between the agent and the environment is always the same loop:

1. `env.reset()` starts an episode and returns the first observation;
2. the agent chooses an action;
3. `env.step(action)` applies the action and returns the new observation, the reward, and whether the episode ended: `terminated` (e.g., the pole fell) or `truncated` (e.g., time limit reached).

Here, the agent chooses random actions in the CartPole environment.

```python
import gymnasium as gym
from gymnasium.utils.save_video import save_video
from IPython.display import Video, display
```

Create the environment. `rgb_array_list` keeps the frames of the episode, to save a video (locally, you can use `render_mode="human"` to watch it in a window).

```python
env = gym.make("CartPole-v1", render_mode="rgb_array_list")
obs, info = env.reset(seed=0)
print("Observation:", obs)
print("Action space:", env.action_space)
```

```python
episode = 0
episode_reward = 0
for _ in range(300):
    # Example: random policy
    action = env.action_space.sample()
    obs, reward, terminated, truncated, info = env.step(action)
    episode_reward += reward
    if terminated or truncated:
        print(f"Episode {episode}: reward {episode_reward}")
        if episode == 0:
            # save the video of the first episode
            save_video(env.render(), "videos", fps=env.metadata["render_fps"], name_prefix="random",
                       episode_trigger=lambda e: True)
        episode += 1
        episode_reward = 0
        obs, info = env.reset()
env.close()
```
